```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
from google.colab import files
import zipfile
import os

uploaded = files.upload()

zip_file = list(uploaded.keys())[0]

print("Uploaded file:", zip_file)

with zipfile.ZipFile(zip_file, 'r') as zip_ref:
    zip_ref.extractall("/content/dataset")

print("ZIP extracted successfully!")
```

```python
import glob

csv_files = glob.glob("/content/dataset/**/*.csv", recursive=True)

print("CSV files found:")

for file in csv_files:
    print(file)
```

```python
csv_path = csv_files[0]

df = pd.read_csv(csv_path)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
display(df.head())
```

```python
print("Column Names:")
print(df.columns.tolist())
```

```python
print("Dataset Information:")
df.info()
```

```python
missing_values = df.isnull().sum()

print("Missing Values:")
print(missing_values)
```

```python
missing_percentage = (
    df.isnull().sum() / len(df) * 100
).sort_values(ascending=False)

display(missing_percentage)
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
df = df.drop_duplicates()

print("Duplicates removed.")
print("New dataset shape:", df.shape)
```

```python
df["date"] = pd.to_datetime(df["date"], errors="coerce")

print(df["date"].head())
print("\nDate range:")
print("Start:", df["date"].min())
print("End:", df["date"].max())
```

```python
display(df.describe())
```

```python
total_demand = df["units_sold"].sum()

print("Total Units Sold / Demand:", total_demand)
```

```python
total_inventory = df["stock_available"].sum()

print("Total Available Inventory:", total_inventory)
```

```python
average_demand = df["units_sold"].mean()

print("Average Demand:", round(average_demand, 2))
```

```python
average_inventory = df["stock_available"].mean()

print("Average Available Inventory:", round(average_inventory, 2))
```

```python
daily_demand = df.groupby("date")["units_sold"].sum()

plt.figure(figsize=(14, 5))

plt.plot(daily_demand)

plt.title("Daily Demand Trend")
plt.xlabel("Date")
plt.ylabel("Units Sold")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
daily_inventory = df.groupby("date")["stock_available"].sum()

plt.figure(figsize=(14, 5))

plt.plot(daily_inventory)

plt.title("Daily Inventory Availability")
plt.xlabel("Date")
plt.ylabel("Available Stock")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
region_demand = (
    df.groupby("region")["units_sold"]
    .sum()
    .sort_values(ascending=False)
)

display(region_demand)
```

```python
plt.figure(figsize=(10, 5))

region_demand.plot(kind="bar")

plt.title("Demand by Region")
plt.xlabel("Region")
plt.ylabel("Units Sold")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
product_demand = (
    df.groupby("product_type")["units_sold"]
    .sum()
    .sort_values(ascending=False)
)

display(product_demand)
```

```python
plt.figure(figsize=(10, 5))

product_demand.plot(kind="bar")

plt.title("Demand by Product Type")
plt.xlabel("Product")
plt.ylabel("Units Sold")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
analysis = df.groupby("region").agg({
    "units_sold": "sum",
    "stock_available": "sum"
}).reset_index()

display(analysis)
```

```python
x = np.arange(len(analysis))
width = 0.35

plt.figure(figsize=(12, 6))

plt.bar(
    x - width/2,
    analysis["units_sold"],
    width,
    label="Demand"
)

plt.bar(
    x + width/2,
    analysis["stock_available"],
    width,
    label="Available Inventory"
)

plt.xlabel("Region")
plt.ylabel("Units")
plt.title("Demand vs Available Inventory by Region")

plt.xticks(x, analysis["region"], rotation=45)

plt.legend()
plt.tight_layout()
plt.show()
```

```python
df["stockout_risk"] = (
    df["units_sold"] > df["stock_available"]
)

stockout_records = df["stockout_risk"].sum()

print("Records with Stockout Risk:", stockout_records)
```

```python
stockout_percentage = (
    df["stockout_risk"].mean() * 100
)

print(
    "Stockout Risk Percentage:",
    round(stockout_percentage, 2),
    "%"
)
```

```python
region_stockout = (
    df.groupby("region")["stockout_risk"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

display(region_stockout)
```

```python
plt.figure(figsize=(10, 5))

region_stockout.plot(kind="bar")

plt.title("Stockout Risk by Region")
plt.xlabel("Region")
plt.ylabel("Stockout Risk (%)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
print("=" * 50)
print("DEMAND & INVENTORY ANALYSIS")
print("=" * 50)

print("Total Records:", len(df))
print("Total Demand:", round(df["units_sold"].sum(), 2))
print("Total Inventory:", round(df["stock_available"].sum(), 2))
print("Average Demand:", round(df["units_sold"].mean(), 2))
print("Average Inventory:", round(df["stock_available"].mean(), 2))
print("Stockout Risk:", round(stockout_percentage, 2), "%")

print("=" * 50)
print("ANALYSIS COMPLETED!")
print("=" * 50)
```
